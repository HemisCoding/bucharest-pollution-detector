```python
import requests
import pandas as pd
from tqdm import tqdm

# Parametrii generali
CITY = "Bucharest"
PARAMETERS = ["pm10", "pm25", "no2"]
DATE_FROM = "2023-01-01"
DATE_TO = "2023-12-31"
LIMIT = 10000  # API-ul permite până la 10.000 pe request

def fetch_openaq_data(parameter):
    print(f"🔄 Fetching {parameter.upper()} data for {CITY} in 2023...")
    url = "https://api.openaq.org/v2/measurements"
    page = 1
    all_results = []

    while True:
        params = {
            "city": CITY,
            "parameter": parameter,
            "date_from": DATE_FROM,
            "date_to": DATE_TO,
            "limit": LIMIT,
            "page": page,
            "sort": "asc",
            "order_by": "datetime"
        }

        response = requests.get(url, params=params)
        data = response.json()

        if "results" not in data or len(data["results"]) == 0:
            break

        all_results.extend(data["results"])
        print(f"📦 Page {page} - Retrieved {len(data['results'])} rows")

        if len(data["results"]) < LIMIT:
            break

        page += 1

    return pd.DataFrame(all_results)

# Descărcăm datele pentru fiecare poluant
dfs = {}
for param in PARAMETERS:
    df = fetch_openaq_data(param)
    if not df.empty:
        df["pollutant"] = param.upper()
        dfs[param] = df

# Combinăm totul într-un singur DataFrame
df_all = pd.concat(dfs.values(), ignore_index=True)

# Păstrăm coloanele utile
df_clean = df_all[["location", "parameter", "value", "unit", "date"]].copy()
df_clean["datetime"] = pd.to_datetime(df_clean["date"].apply(lambda d: d["utc"]))
df_clean.drop(columns=["date"], inplace=True)

print("✅ Date extrase:")
print(df_clean.head())

# Salvează opțional în CSV
df_clean.to_csv("bucharest_air_quality_2023.csv", index=False)
```
